EXPLORATORY DATA ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
data=pd.read_csv("C:\\leadsense-ai\\backend\\data\\synthetic_leads.csv")
print("data shape:",data.shape)
```

```python
#Conversion distribution
plt.figure()
sns.countplot(x="converted",data=data)
plt.title("Lead Conversion Distribution")
plt.show()
```

```python
#Income vs Conversion
plt.figure()
sns.boxplot(x="converted",y="income",data=data)
plt.title("Income vs Conversion")
plt.show()
```

```python
#Credit score vs conversion
plt.figure()
sns.boxplot(x="converted",y="credit_score",data=data)
plt.title("Credit score vs Conversion")
plt.show()
```

```python
#Traffic Source Impact
plt.figure()
sns.countplot(x="traffic_source",hue="converted",data=data)
plt.title("Traffic Source vs Conversion")
plt.show()
```

```python
#engagement vs conversion
plt.figure()
sns.scatterplot(x="pages_viewed",y="time_on_site",hue="converted",data=data)
plt.title("Engagement vs Conversion")
plt.show()
```
